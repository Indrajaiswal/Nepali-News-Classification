```python
news_df = df[["heading", "text", "clean_categories"]].copy()

news_df.rename(
    columns={
        "text": "news_body",
        "clean_categories": "category"
    },
    inplace=True
)

news_df.head()
```

```python
news_df.dropna(
    subset=["news_body","category"],
    inplace=True
)

news_df = news_df[
    news_df["news_body"].str.strip() != ""
]
```

```python
news_df.drop_duplicates(
    subset=[ "news_body"],
    inplace=True
)
```

```python
print(news_df.shape)
```

```python
category_counts = (
    news_df["category"]
    .value_counts()
    .sort_values(ascending=False)
)

print(category_counts)
```

```python
selected_categories = [
    "politics",
    "national",
    "society",
    "economy",
    "sports",
    "global",
    "health",
    "entertainment",
    "business",
    "education",
    "science and technology",
    "crime"
]

news_df = news_df[
    news_df["category"].isin(selected_categories)
]
```

```python
news_df["category"].value_counts()
```

```python
sample_list = []

for category, group in news_df.groupby("category"):
    sample = group.sample(
        n=8000,
        random_state=42
    )
    sample_list.append(sample)

sample_df = pd.concat(
    sample_list,
    ignore_index=True
)

print(sample_df.columns)
print(sample_df.shape)
```

```python
import re
import pandas as pd

def clean_text(text):
    if pd.isna(text):
        return text

    # Remove unusual Unicode line separators
    text = re.sub(r'[\u2028\u2029]', ' ', text)

    # Replace newlines and tabs with spaces
    text = re.sub(r'[\r\n\t]+', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

sample_df["news_body"] = sample_df["news_body"].apply(clean_text)
sample_df["heading"] = sample_df["heading"].apply(clean_text)
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

sample_df.to_csv(
    "../data/processed/sample_dataset.csv",
    index=False,
    encoding="utf-8-sig"
)
```
